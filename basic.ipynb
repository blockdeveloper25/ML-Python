```python
prices = [7,1,5,3,6,4]

min_price = float('inf')
max_profit = 0

for price in prices:
    if price < min_price:
        min_price = price
    else:
        max_profit = max(max_profit, price - min_price)

print(max_profit)
```

```python

min_price = float('inf')
max_profit = 0
for price in prices:
    if price < min_price:
        min_price = price
    
    profit = price - min_price
    if profit > max_profit:
        max_profit = profit
print(max_profit)
       
```

```python
def mergeAlternately(self, word1: str, word2: str) -> str:
        merged = []
        i = j = 0
        len1, len2 = len(word1), len(word2)

        while i < len1 or j < len2:
            if i < len1:
                merged.append(word1[i])
                i += 1
            if j < len2:
                merged.append(word2[j])
                j += 1
        
        return ''.join(merged)

    
```

```python
from collections import Counter

letter1 = 'ABABAB'
letter2 = 'ABAB'
res = []
# Count letters in both strings
count1 = Counter(letter1)
count2 = Counter(letter2)

# Find common letters
common = set(letter1) & set(letter2)

# Check which common letters appear more than once in either string
for letter in common:
    if count1[letter] > 1 or count2[letter] > 1:
        res.append(letter)

print(''.join(res))
```

```python
def gcdOfStrings(str1: str, str2: str) -> str:
        if str1 + str2 != str2 + str1:
            return ""
    
        gcd_len = math.gcd(len(str1), len(str2))
            
        return str1[:gcd_len]
```

```python
candies = [2,3,5,1,3]
extraCandies = 3

# Step 1: Find the current maximum candies any kid has
max_candies = max(candies)

# Step 2: Compare each child's candies + extra to the max
res = []
for c in candies:
    res.append((c + extraCandies) >= max_candies)

print(res)
```

```python
def canPlaceFlowers(flowerbed, n):
    i = 0
    length = len(flowerbed)
    
    while i < length:
        if flowerbed[i] == 0:
            empty_left = (i == 0 or flowerbed[i - 1] == 0)
            empty_right = (i == length - 1 or flowerbed[i + 1] == 0)

            if empty_left and empty_right:
                flowerbed[i] = 1
                n -= 1
                if n == 0:
                    return True
                i += 2  
                continue
        i += 1
    return n <= 0
```

```python
print(canPlaceFlowers([1, 0, 0, 0, 1], 1))  # ✅ True
print(canPlaceFlowers([1, 0, 0, 0, 1], 2))  # ❌ False
```

```python
class Solution:
    def canPlaceFlowers( flowerbed, n):
        length = len(flowerbed)
        if length ==1 and n==1:
            return True if n==1 and flowerbed[0]==0 else False
        if length ==1 and n==0:
            return True
        for i in range(length):
            if n ==0:
                break
            if flowerbed[i]==1:
                i+=2
                continue
            if i==0 and flowerbed[i+1]==0:
                n-=1
                flowerbed[i]=1
                continue
            if i==length-1 and flowerbed[i-1]==0:
                n-=1
                flowerbed[i]=1
                continue
            if flowerbed[i-1]==0 and flowerbed[i+1]==0:
                n-=1
                flowerbed[i]=1
        return True if n == 0 else False
```
